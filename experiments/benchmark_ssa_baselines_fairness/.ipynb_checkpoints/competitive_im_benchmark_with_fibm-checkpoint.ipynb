## 1. Research overview

Competitive influence maximization studies how a campaign (Rumor A / truth) should choose seed nodes against an adversarial campaign (Rumor B) on a network. This project uses a **signed trust network**:

- the CSV records trust arcs `source → target`;
- the influence graph is the exact reverse, `target → source`;
- positive trust arcs preserve the rumor;
- negative trust arcs flip the rumor;
- transmission probability is determined by the absolute trust strength.

The experiment compares three SSA objective variants in **SSA-only** mode against the seed-selection algorithms from the earlier notebook, plus a fairness-aware influence-blocking baseline from the research literature. The comparison is deliberately not a single-objective optimization contest: SSA optimizes its own fitness, CELF/NCGE use the notebook's O1 fitness, and FIBM optimizes maximin fairness on a CIC surrogate. All methods are nevertheless evaluated with the same external Net Advantage protocol.

The notebook is organized to be readable and executable cell by cell. The expensive work has already been done and is reused.

## 3. Experimental methodology

### 3.1 Network and influence direction

The trust CSV has columns `source, target, rating, probability, strength, polarity`. The influence graph reverses every arc exactly once. All algorithms and the diffusion engine use the reversed graph.

### 3.2 Diffusion model

The evaluated model is a finite-contact competitive Maki–Thompson variant:

- states: `0` ignorant, `1` A-spreader, `2` B-spreader, `3` A-stifler, `4` B-stifler;
- each spreader contacts one uniformly random uncontacted outgoing influence neighbour per round;
- no uncontacted neighbour, or a non-ignorant target, makes the spreader a stifler;
- otherwise it transmits with the edge probability;
- a negative edge flips A↔B; a positive edge preserves the rumor;
- proposals are resolved by highest edge strength; different-rumor ties are random;
- `Spread_A` counts states 1+3; `Spread_B` counts 2+4.

### 3.3 Opponents and budgets

- B1 Random, B2 Degree, B3 Positive PageRank, B4 Community-concentrated.
- B orders have length 50 and are nested across K; the K-seed set is the prefix.
- K = 10, 20, 30, 40, 50.
- Optimization MC = 15 common seeds; final evaluation MC = 100 common seeds.

### 3.4 Primary metric

```
Net Advantage = Spread_A − Spread_B
```

The ranking is by Net Advantage. `Spread_A` and `Spread_B` are retained to explain why a method wins or loses. A method is not declared best from `Spread_A` alone or from a blocking metric alone.

## 10. SSA O1, O4 and O6

The SSA implementation is a discrete set-based Salp Swarm Algorithm:

- population 12, iterations 10, mutation 0.20;
- `c1 = 2·exp(−(4t/T)²)`, exploitation probability `1 − c1/2`;
- leader/follower updates, repair to exactly K unique candidate-pool nodes;
- best-so-far food preserved; worst individual replaced by food;
- fitness = mean objective over 15 optimization Monte Carlo seeds;
- candidate pool = 10·K, 70% ranked / 30% random from the ranked remainder.

| ID | Formula | K-dependent coefficients |
|---|---|---|
| O1 | `Spread_A + 1.5·Blocked_B` | none |
| O4 | `Spread_A + 2.0·Blocked_B` | none |
| O6 | `α(K)·Spread_A + β(K)·Blocked_B` | K≤20: (0.75, 2.25); K≤40: (1.00, 1.75); K=50: (1.25, 1.50) |

### 10.1 SSA O1 — Original SSA

O1 preserves the original notebook fitness: `Spread_A + 1.5·Blocked_B`. It rewards reaching A-susceptible nodes and reducing B spread, with a fixed 1.5 weight on blocking. It does not subtract `Spread_B` explicitly; the blocking term is the paired B-only baseline minus the observed B spread.

### 10.2 SSA O4 — Combined influence-blocking

O4 uses `Spread_A + 2.0·Blocked_B`. It increases the blocking weight relative to O1 while keeping the same SSA mechanics and candidate pool. O4 is distinct from O5 (which subtracts `Spread_B` explicitly); O4 is the version retained here.

### 10.3 SSA O6 — Budget-aware

O6 uses `α(K)·Spread_A + β(K)·Blocked_B` with K-dependent coefficients: K≤20 uses (0.75, 2.25), K≤40 uses (1.00, 1.75), and K=50 uses (1.25, 1.50). Small budgets emphasize blocking; large budgets emphasize A reach while still rewarding blocking.

`Blocked_B = max(0, mean_s baseline_B(B,s) − mean_s Spread_B(A,B,s))`, with paired MC seeds.

SSA is run **once** per `(scenario, K, objective)`. The saved O1/O4/O6 seed sets are reused; the notebook does not rerun SSA in load-existing mode.


````python
SSA_METHOD_IDS = ["SSA_O1", "SSA_O4", "SSA_O6"]
SSA_OBJECTIVE = {"SSA_O1": "O1", "SSA_O4": "O4", "SSA_O6": "O6"}

def ssa_variant_index(objective_id):
    return list(VARIANTS.keys()).index(objective_id)

def ssa_seed_set_for(method_id, scenario, k):
    """Return the saved SSA seed set for one cell, without running SSA."""
    results = pd.read_csv(EXPERIMENT_DIR / "raw_results" / "results.csv")
    row = results[
        (results["method_id"] == method_id)
        & (results["scenario"] == scenario)
        & (results["K"].astype(int) == int(k))
    ]
    if len(row) != 1:
        raise ValueError(f"Expected one saved row for {method_id}/{scenario}/K={k}")
    return sorted(json.loads(row.iloc[0]["seed_set"]))

# Show the SSA budget and the O6 coefficients; no search is launched.
ssa_budget = pd.DataFrame([
    {"parameter": "population", "value": config["ssa"]["population"]},
    {"parameter": "iterations", "value": config["ssa"]["iterations"]},
    {"parameter": "mutation rate", "value": config["ssa"]["mutation_rate"]},
    {"parameter": "objective evaluations per run", "value": 12 + 10 * 12},
    {"parameter": "optimization MC seeds", "value": config["optimization_mc_runs"]},
    {"parameter": "candidate pool ratio", "value": config["ssa"]["pool_ratio"]},
    {"parameter": "random pool fraction", "value": config["ssa"]["random_pool_fraction"]},
])
display(ssa_budget)

o6_rows = []
for k in config["k_values"]:
    alpha, beta = budget_coefficients(k)
    o6_rows.append({"K": k, "alpha(K)": alpha, "beta(K)": beta})
display(pd.DataFrame(o6_rows))
````

## 11. Traditional baseline algorithms

The traditional and literature baselines are the actual notebook implementations:

- **Random** — uniform sample of eligible nodes.
- **Top Degree** — influence out-degree ranking.
- **Signed PageRank** — two-layer signed random walk; transition weight = absolute rating; teleport to the positive layer only.

## 12. FIBM fairness baseline

**Full name:** *Fairness-Aware Influence Blocking Maximization based on Martingales* (**FIBMM**, referred to in this notebook as **FIBM**).  
**Paper:** Xiao, Liu, Yang and Li, “Fairness-Aware Influence Blocking Maximization in Social Networks,” *IEEE Transactions on Computational Social Systems*, 12(6), 2025.

### What the paper does

The paper defines the Fairness Influence Blocking Maximization (FIBM) problem on an unsigned Competitive Independent Cascade (CIC) network:

```
sigma(S) = (1 - h) * Pi(S) + h * tau(S)
```

- `Pi(S)` is the overall saving ratio (protected nodes / nodes that would be infected without truth seeds);
- `tau(S)` is the minimum community saving ratio (maximin fairness);
- `h` controls the utility/fairness trade-off;
- the method samples community-aware reverse protector (CRP) sets and uses a two-stage greedy selection:
  1. `k - floor(k*h)` nodes by overall marginal coverage;
  2. the remaining nodes by repeatedly choosing the currently worst-off community and the node that covers most of its remaining CRP sets.

The paper's defaults used here are `h=0.9`, `ε=0.1`, `l=1`, Louvain communities, and a small-community merge threshold of `0.005·n`.

### Adaptation to our signed competitive Maki–Thompson model

The paper assumes an **unsigned CIC** model. Our evaluated model is a **signed finite-contact Maki–Thompson** model. The selection step therefore uses the minimum defensible surrogate:

- **positive-edge CIC surrogate** for FIBM's CRP sampling; negative edges are excluded because they flip the rumor and cannot reliably carry the truth cascade;
- Louvain communities on the undirected unweighted projection of **all** influence edges (signs ignored), matching the project's B4 convention;
- the paper's martingale sample-size formula is implemented, but the total CRP budget is capped at **600,000 per configuration** for feasibility; raw and capped sizes are recorded;
- final performance is always measured with the **full signed Maki–Thompson** evaluator, not the surrogate;
- FIBM is **not** rewritten into O1, O4, or O6. Its own selection logic is preserved.

This is an adaptation, not a claim of identity with the paper's unsigned-CIC experiments. The notebook exposes both the internal fairness estimates and the external Net Advantage/Blocked_B results.


````python
FIBM_METHOD_ID = "FIBMM"   # internal method id used by the saved benchmark

def fibm_config_from_experiment(scenario, k):
    """Build the paper-based FIBM configuration for one cell."""
    fib = config["fibmm"]
    n = int(scenario[1:])
    seed = int(config["master_seed"]) + int(fib["crp_seed_offset"]) + n * 1000 + int(k)
    return FIBMMConfig(
        h=float(fib["h"]),
        epsilon=float(fib["epsilon"]),
        l=int(fib["l"]),
        community_min_fraction=float(fib["community_min_fraction"]),
        community_seed=int(fib["community_seed"]),
        max_crp_samples=int(fib["max_crp_samples_per_config"]),
        min_samples_per_community=int(fib["min_samples_per_community"]),
        crp_seed=seed,
    )

def fibm_saved_result(scenario="B1", k=10):
    row = results_df[
        (results_df["method_id"] == FIBM_METHOD_ID)
        & (results_df["scenario"] == scenario)
        & (results_df["K"].astype(int) == int(k))
    ]
    if len(row) != 1:
        raise ValueError(f"Expected one saved FIBM row for {scenario}/K={k}")
    return row.iloc[0]

# Show the FIBM configuration and the saved B1/K=10 external result.
fibm_cfg = fibm_config_from_experiment("B1", 10)
fibm_row = fibm_saved_result("B1", 10)
display(pd.DataFrame([{
    "method": "FIBM",
    "internal_id": FIBM_METHOD_ID,
    "scenario": "B1",
    "K": 10,
    "h": fibm_cfg.h,
    "epsilon": fibm_cfg.epsilon,
    "l": fibm_cfg.l,
    "crp_cap": fibm_cfg.max_crp_samples,
    "net_advantage": fibm_row["net_advantage"],
    "spread_A": fibm_row["spread_A"],
    "spread_B": fibm_row["spread_B"],
    "blocked_B": fibm_row["blocked_B"],
}]))

# Load the saved FIBM diagnostics; no CRP sampling is launched.
diag_path = EXPERIMENT_DIR / "raw_results" / "fibmm_diagnostics" / "FIBMM_B1_K10.json"
fibm_diag = json.loads(diag_path.read_text(encoding="utf-8"))
display(pd.DataFrame([{
    "n_communities": fibm_diag["n_communities"],
    "crp_sets": fibm_diag["crp_sets"],
    "pi_total": fibm_diag["pi_total"],
    "tau": fibm_diag["tau"],
    "sigma": fibm_diag["sigma"],
    "stage1_count": fibm_diag["stage1_count"],
    "stage2_count": fibm_diag["stage2_count"],
    "theta_raw_total": fibm_diag["sample_plan"]["theta_raw_total"],
    "theta_total_capped": fibm_diag["sample_plan"]["theta_total"],
}]))

display(pd.DataFrame(fibm_diag["community_metrics"]).head(10))
````

### FIBM compared with the SSA variants

FIBM is not expected to maximize Net Advantage: its internal objective is maximin fairness on a CIC surrogate. The comparison below is descriptive and uses the same 100 external evaluation seeds as every other method.


````python
fibm_compare_rows = []
for ssa in SSA_METHOD_IDS:
    diffs_net, diffs_blocked = [], []
    wins = ties = losses = 0
    for sid in config["scenarios"]:
        for k in config["k_values"]:
            a = results_df[
                (results_df["method_id"] == ssa)
                & (results_df["scenario"] == sid)
                & (results_df["K"] == k)
            ].iloc[0]
            b = results_df[
                (results_df["method_id"] == FIBM_METHOD_ID)
                & (results_df["scenario"] == sid)
                & (results_df["K"] == k)
            ].iloc[0]
            d = float(a["net_advantage"] - b["net_advantage"])
            db = float(a["blocked_B"] - b["blocked_B"])
            diffs_net.append(d)
            diffs_blocked.append(db)
            if abs(d) <= 1e-9:
                ties += 1
            elif d > 0:
                wins += 1
            else:
                losses += 1
    fibm_compare_rows.append({
        "SSA": ssa,
        "wins": wins,
        "ties": ties,
        "losses": losses,
        "mean_net_advantage_difference": float(np.mean(diffs_net)),
        "mean_blocked_B_difference": float(np.mean(diffs_blocked)),
    })
display(pd.DataFrame(fibm_compare_rows))
````

## 14. Experiment execution, persistence and reuse

The completed experiment is stored under `raw_results/`. The notebook checks:

1. the expected 4 × 5 × 12 = 240 cells;
2. the current configuration signature for every row;
3. the 100-row common-seed MC file for every row;
4. the ledger and manifest status.

If a cell is missing or invalid, the notebook **does not silently compute it**. The default mode reports it. `run_missing` requires `ALLOW_MISSING_RUNS=True`; `force_rerun` writes to a new timestamped directory and requires explicit method/scenario/K selections. Existing saved results are never overwritten by the notebook's force path.


````python
def _expected_cells(config):
    return [
        (m["id"], sid, int(k))
        for m in config["methods"]
        for sid in config["scenarios"]
        for k in config["k_values"]
    ]

def check_result_reuse():
    results_path = EXPERIMENT_DIR / "raw_results" / "results.csv"
    ledger_path = EXPERIMENT_DIR / "raw_results" / "ledger.csv"
    manifest_path = EXPERIMENT_DIR / "raw_results" / "manifest.json"

    results = pd.read_csv(results_path)
    results["K"] = results["K"].astype(int)
    ledger = pd.read_csv(ledger_path) if ledger_path.exists() else pd.DataFrame()
    manifest = json.loads(manifest_path.read_text(encoding="utf-8")) if manifest_path.exists() else {}

    expected = set(_expected_cells(config))
    present = {(r.method_id, r.scenario, int(r.K)) for r in results.itertuples(index=False)}
    missing = sorted(expected - present)
    extra = sorted(present - expected)

    signature_mismatches = []
    for row in results.itertuples(index=False):
        B = json.loads(row.opponent_seed_set)
        expected_sig = cell_signature(
            config, row.method_id, row.scenario, int(row.K), B, network_meta
        )
        if str(row.config_signature_hash) != expected_sig:
            signature_mismatches.append((row.method_id, row.scenario, int(row.K)))

    mc_bad = []
    for row in results.itertuples(index=False):
        mc_path = EXPERIMENT_DIR / "raw_results" / str(row.mc_file)
        if not mc_path.exists():
            mc_bad.append((row.method_id, row.scenario, int(row.K), "missing"))
            continue
        mc = pd.read_csv(mc_path)
        if len(mc) != int(config["final_mc_runs"]):
            mc_bad.append((row.method_id, row.scenario, int(row.K), f"{len(mc)} rows"))
        elif list(mc["seed"].astype(int)) != list(range(900000000, 900000000 + int(config["final_mc_runs"]))):
            mc_bad.append((row.method_id, row.scenario, int(row.K), "seed list"))

    b_counts = results.groupby(["scenario", "K"])["opponent_seed_set"].apply(
        lambda values: len({tuple(sorted(json.loads(v))) for v in values})
    )
    b_inconsistent = b_counts[b_counts != 1]

    status_counts = results["status"].value_counts().to_dict()
    ledger_counts = ledger["status"].value_counts().to_dict() if not ledger.empty else {}
    return {
        "results": results,
        "ledger": ledger,
        "manifest": manifest,
        "expected_cells": len(expected),
        "present_cells": len(present),
        "missing_cells": missing,
        "extra_cells": extra,
        "signature_mismatches": signature_mismatches,
        "mc_bad": mc_bad,
        "b_inconsistent": b_inconsistent,
        "status_counts": status_counts,
        "ledger_counts": ledger_counts,
        "manifest_completed": manifest.get("completed_cells"),
        "manifest_expected": manifest.get("expected_cells"),
    }

reuse_audit = check_result_reuse()
display(pd.DataFrame([{
    "expected_cells": reuse_audit["expected_cells"],
    "present_cells": reuse_audit["present_cells"],
    "missing_cells": len(reuse_audit["missing_cells"]),
    "extra_cells": len(reuse_audit["extra_cells"]),
    "signature_mismatches": len(reuse_audit["signature_mismatches"]),
    "bad_mc_files": len(reuse_audit["mc_bad"]),
    "opponent_inconsistent_cells": len(reuse_audit["b_inconsistent"]),
    "manifest_completed": reuse_audit["manifest_completed"],
    "manifest_expected": reuse_audit["manifest_expected"],
}]))
display(pd.DataFrame([reuse_audit["status_counts"]]))
display(pd.DataFrame([reuse_audit["ledger_counts"]]))
if (reuse_audit["missing_cells"] or reuse_audit["signature_mismatches"]
        or reuse_audit["mc_bad"] or len(reuse_audit["b_inconsistent"])):
    print("Reuse audit found issues; inspect the lists above before running anything.")
else:
    print("Reuse audit: all 240 cells are present, signatures match, MC files are valid, "
          "and opponent B sets are identical within every scenario/K cell.")
````

````python
def execute_requested_work():
    """Guarded execution path.  Default load-existing mode is a no-op."""
    global SIMULATION_LAUNCHED
    if EXECUTION_MODE == "load_existing":
        print("Execution mode 'load_existing': no SSA, FIBM, baseline or diffusion run launched.")
        return

    if EXECUTION_MODE == "run_missing":
        if not ALLOW_MISSING_RUNS:
            raise RuntimeError("Set ALLOW_MISSING_RUNS=True to run missing cells.")
        missing = reuse_audit["missing_cells"] + [
            (m, s, k) for (m, s, k) in reuse_audit["signature_mismatches"]
        ]
        if not missing:
            print("No missing or invalid cells. Nothing to run.")
            return
        from run_benchmark import BenchmarkRunner
        SIMULATION_LAUNCHED = True
        runner = BenchmarkRunner(config, output_root=EXPERIMENT_DIR / "raw_results")
        for method_id, scenario, k in missing:
            runner.run_cell(method_id, scenario, int(k), force=False)
        print(f"Completed {len(missing)} missing/invalid cells.")
        return

    if EXECUTION_MODE == "force_rerun":
        if not ALLOW_FORCE_RERUN:
            raise RuntimeError("Set ALLOW_FORCE_RERUN=True for a force rerun.")
        if not (FORCE_METHODS and FORCE_SCENARIOS and FORCE_K):
            raise RuntimeError(
                "Force rerun requires explicit FORCE_METHODS, FORCE_SCENARIOS and FORCE_K."
            )
        from run_benchmark import BenchmarkRunner
        stamp = time.strftime("%Y%m%d_%H%M%S")
        out = EXPERIMENT_DIR / "notebook_reruns" / stamp
        SIMULATION_LAUNCHED = True
        runner = BenchmarkRunner(
            config, output_root=out,
            only_methods=FORCE_METHODS,
            only_scenarios=FORCE_SCENARIOS,
            only_k=FORCE_K,
        )
        runner.run_all(force=True)
        print("Force rerun written to:", out)
        return

    raise ValueError(f"Unknown EXECUTION_MODE: {EXECUTION_MODE}")

execute_requested_work()
````

## 22. Visualizations

The charts use the cleaner notebook style: two logical panels per scenario, consistent Okabe–Ito-derived colors, distinct markers and line styles, 95% Monte Carlo error bars, a shared y-axis, and a bottom legend. The left panel contains the SSA objectives, SSA-Notebook and FIBM; the right panel contains the traditional/literature baselines.

The figures are written to `figures/notebook/` so the existing final figures are not overwritten. No simulation is launched.


````python
# Use the analysis module's consistent style constants.
COLORS = dict(ab.METHOD_COLORS)
MARKERS = dict(ab.METHOD_MARKERS)
LINESTYLES = dict(ab.METHOD_LINESTYLES)
LABELS = {m: ab.METHOD_LABELS[m].replace("FIBMM", "FIBM") for m in ab.METHOD_ORDER}
FOCUS = ["SSA_O1", "SSA_O4", "SSA_O6", "SSA_Notebook", "FIBMM"]
BASELINES = ["Random", "TopDegree", "SignedPageRank", "CRIEG", "NCGE", "SignedGA", "CELF"]

def _plot_panel(ax, data, methods, scenario):
    for method_id in methods:
        sub = data[
            (data["method_id"] == method_id) & (data["scenario"] == scenario)
        ].sort_values("K")
        if sub.empty:
            continue
        y = sub["net_advantage"].to_numpy(float)
        lo = sub["net_advantage_ci_low"].to_numpy(float)
        hi = sub["net_advantage_ci_high"].to_numpy(float)
        yerr = np.vstack([y - lo, hi - y])
        lw = 2.8 if method_id == "FIBMM" else (2.3 if method_id in SSA_METHOD_IDS else 1.8)
        ms = 9 if method_id == "FIBMM" else (6.5 if method_id in SSA_METHOD_IDS else 5.5)
        ax.errorbar(
            sub["K"].to_numpy(int), y, yerr=yerr,
            color=COLORS[method_id], marker=MARKERS[method_id],
            linestyle=LINESTYLES[method_id], linewidth=lw, markersize=ms,
            capsize=2.5, elinewidth=0.9, alpha=0.92,
            label=LABELS[method_id],
            zorder=3 if method_id in SSA_METHOD_IDS + ["FIBMM"] else 2,
        )
    ax.axhline(0, color="#333333", linestyle="--", linewidth=0.9, alpha=0.55, zorder=1)
    ax.set_xticks(config["k_values"])
    ax.set_xticklabels([str(k) for k in config["k_values"]])
    ax.set_xlabel("Seed budget K")
    ax.grid(True, which="major", alpha=0.25, linewidth=0.7)

def plot_scenario(sid, data, outdir):
    fig, axes = plt.subplots(1, 2, figsize=(16.8, 7.0), sharey=True)
    _plot_panel(axes[0], data, FOCUS, sid)
    _plot_panel(axes[1], data, BASELINES, sid)
    axes[0].set_title("SSA objectives, notebook SSA and FIBM", fontsize=12, fontweight="bold")
    axes[1].set_title("Traditional and literature baselines", fontsize=12, fontweight="bold")
    axes[0].set_ylabel("Net Advantage (Spread_A − Spread_B)")

    all_lo = min(float(data[data["scenario"] == sid]["net_advantage_ci_low"].min()),
                 float(data[data["scenario"] == sid]["net_advantage"].min()))
    all_hi = max(float(data[data["scenario"] == sid]["net_advantage_ci_high"].max()),
                 float(data[data["scenario"] == sid]["net_advantage"].max()))
    pad = max(5.0, 0.08 * (all_hi - all_lo))
    axes[0].set_ylim(all_lo - pad, all_hi + pad)

    handles, labels = [], []
    for ax in axes:
        h, l = ax.get_legend_handles_labels()
        handles += h
        labels += l
    order = {LABELS[m]: i for i, m in enumerate(ab.METHOD_ORDER)}
    pairs = sorted(zip(labels, handles), key=lambda x: order.get(x[0], 99))
    labels = [p[0] for p in pairs]
    handles = [p[1] for p in pairs]

    fig.legend(
        handles, labels, loc="lower center", ncol=6, frameon=False,
        bbox_to_anchor=(0.5, 0.055), fontsize=8.8, handlelength=2.4,
        columnspacing=1.3,
    )
    fig.suptitle(
        f"{sid} — {ab.SCENARIO_TITLES[sid]}: Net Advantage vs K",
        fontsize=15, fontweight="bold", y=0.975,
    )
    fig.text(
        0.5, 0.012,
        "Error bars: 95% CI over 100 common Monte Carlo seeds (evaluation uncertainty). "
        "One optimizer run per method/configuration; bars do not represent optimizer variability.",
        ha="center", fontsize=8.3, color="#555555",
    )
    fig.subplots_adjust(left=0.06, right=0.985, top=0.88, bottom=0.235, wspace=0.06)
    path = outdir / f"Net_Advantage_{sid}.png"
    fig.savefig(path, dpi=300, bbox_inches="tight", facecolor="white")
    plt.close(fig)
    return path

notebook_figure_paths = []
for sid in config["scenarios"]:
    notebook_figure_paths.append(plot_scenario(sid, results_df, NOTEBOOK_FIG_DIR))

for path in notebook_figure_paths:
    print(path)
    display(Image(filename=str(path), width=1200))
````

## 23. Findings and interpretation

### Overall

- **SSA O1** has the highest mean Net Advantage: **47.878**.
- **SSA O6** is second: **47.261**.
- **CELF** is the strongest non-SSA baseline: **46.115**.
- **SSA-Notebook** is **45.570**; **SSA O4** is **43.241**.
- CELF and SSA O1 are tied for the most strict wins: **6/20 each**.
- There is **no universal winner**. The winner changes with the opponent scenario and K.

### SSA vs the traditional baselines

All three SSA variants beat Random, Top Degree, Signed PageRank and NCGE in every cell. They beat CRIEG in 19–20 of 20 cells and Signed-GA in 17–19 of 20. The real contest is against CELF:

- O1 vs CELF: **10W/10L**
- O6 vs CELF: **10W/10L**
- O4 vs CELF: **9W/11L**

CELF wins B1 K=30/40, B2 K=10, B3 K=20, B4 K=10 and B4 K=50. Its highest cell is B4/K=50 with Net Advantage **92.68**.

### Why CELF is competitive

CELF uses the same weighted candidate pool and O1 fitness as SSA O1, but lazy greedy evaluates many more candidates (about 633 objective evaluations per cell on average, versus 132 for SSA). Its mean is comparable to SSA O1, but its selection runtime is about **4.35 s per cell** versus **0.68 s** for SSA. The result is a genuine cost-performance trade-off, not a free win.

### FIBM

FIBM's mean Net Advantage is **18.718**, far below the SSA variants. Its mean Blocked_B is **34.063**, slightly higher than SSA O1's **33.379**, so it does improve blocking on average. Its mean Spread_A is much lower (965.72 versus 995.56), which is why its Net Advantage is lower. Internal FIBM fairness estimates (Π and τ) are slightly negatively correlated with external Net Advantage and slightly positively correlated with Blocked_B. This is consistent with a fairness-vs-competitive-advantage trade-off.

FIBM is a defensible fairness baseline for the paper's objective, but it is not a Net-Advantage optimizer and should not be expected to win the competitive ranking.

### Uncertainty

Only **4 of 20** winner-vs-runner-up paired 95% intervals exclude zero. Most per-cell margins are within Monte Carlo uncertainty. The per-cell rankings are descriptive, not decisive significance claims.

### Runtime

SSA is the cheapest strong method. CELF and NCGE are moderate. Signed-GA and FIBM are much more expensive per configuration. The benchmark records these differences explicitly.


## 24. Limitations

1. **One optimizer run per method/configuration.** The 100 final Monte Carlo runs measure evaluation uncertainty, not between-optimizer-run variability.
2. **B1–B4 are not replicates.** They are four deliberately different opponent strategies; no pooled significance test across scenarios is claimed.
3. **Synthetic network.** Results are conditional on this 5,000-node synthetic design and the reversed influence convention.
4. **FIBM adaptation.** FIBM's selection uses a positive-edge CIC surrogate because the paper's model is unsigned CIC; final evaluation uses the signed finite-contact Maki–Thompson model. The CRP sample budget is capped at 600,000 per configuration.
5. **CELF and NCGE use O1 fitness**, not Net Advantage. This is faithful to the notebook code but means their internal objective differs from the ranking metric.
6. **Literature adaptations.** CRIEG, NCGE and Signed-GA were adapted to this signed competitive setting; their original papers assume different diffusion models.
7. **CELF + Local Search is excluded** because it is a swap-based post-processing step, per the no-swap instruction. The base CELF algorithm is included.
8. **No claim of universal superiority.** CELF ties SSA O1 in strict wins, and the winner changes by scenario and K.


## 25. Conclusion and recommendation

For the next research stage, prioritize **SSA O1** if the goal is the best average Net Advantage across the current B1–B4 design. Keep **SSA O6** as the close second and **CELF** as the primary cheap baseline.

Do **not** treat SSA O4 as an improvement over O1 on this evidence: its mean is 43.241, it trails O1 by about 4.64, and it loses 9 of 20 cells to CELF.

The next experiments should:

1. repeat the SSA optimizer with independent random seeds to estimate optimizer variability;
2. investigate why CELF wins B1 K=30/40, B2 K=10, B3 K=20, B4 K=10 and B4 K=50;
3. test whether the O1/O4/O6/CELF ranking is stable on a second network realization;
4. retain FIBM as a fairness baseline, not as a replacement for the Net Advantage objective unless fairness is an explicit deployment constraint.
